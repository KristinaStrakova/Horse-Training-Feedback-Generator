
##  Setting up the Ollama

###  1. **Install Ollama**

For **Windows**, install from: [https://ollama.com/download](https://ollama.com/download)

---

###  2. **Download a Model** (In Terminal not in VS Code terminal)

Once installed, still in terminal:

```bash
ollama pull mistral
```

This downloads the **Mistral model** to run locally (you can use `llama2`, `codellama`, or others too).

---

###  3. **Test the Model Interactively** (Terminal)

Try it to make sure it works:

```bash
ollama run mistral
```

It’ll start a local chatbot. Type something like:

```
Horse's ACWR has been above 1.5 for 4 days. Suggest training advice.
```
or
```
Tell me a short joke about horses.
```

If it answers, you’re good to go and Python script should run for you!
